```
using System;
using System.Collections.Generic;
using System.Linq;

// ДЕЛЕГАТЫ
public delegate void PersonEventHandler(string message);
public delegate void WorkEventHandler(string task, int hours);
public delegate void DisplayInfoDelegate();

public interface IPersonalInfo
{
    void DisplayBasicInfo();
    void UpdateContact(string email, string phone);
}

public interface IProfessional
{
    void DisplayProfessionalDetails();
    void PerformWork();
}

public class PersonalInfoService
{
    private readonly IPersonalInfo _person;

    // ДЕЛЕГАТЫ
    public DisplayInfoDelegate OnInfoDisplay;
    public Action<string> OnContactUpdate;

    public PersonalInfoService(IPersonalInfo person)
    {
        _person = person;
        
        // Инициализация делегатов
        OnInfoDisplay = () => Console.WriteLine("Информация будет отображена");
        OnContactUpdate = (message) => Console.WriteLine($"Контакты обновляются: {message}");
    }

    public void ProcessPersonalInfo()
    {
        Console.WriteLine("=== СЕРВИС ПЕРСОНАЛЬНОЙ ИНФОРМАЦИИ ===");
        OnInfoDisplay?.Invoke();
        _person.DisplayBasicInfo();
    }

    public void UpdateContacts(string email, string phone)
    {
        OnContactUpdate?.Invoke($"Email: {email}, Phone: {phone}");
        _person.UpdateContact(email, phone);
    }
}

public class ProfessionalService
{
    private readonly IProfessional _professional;

    public ProfessionalService(IProfessional professional)
    {
        _professional = professional;
    }

    public void ProcessProfessionalInfo()
    {
        Console.WriteLine("=== СЕРВИС ПРОФЕССИОНАЛЬНОЙ ИНФОРМАЦИИ ===");
        _professional.DisplayProfessionalDetails();
        _professional.PerformWork();
    }
}

// КЛАСС Person С СОБЫТИЯМИ
public class Person : IPersonalInfo
{
    public string Name { get; set; }
    public int Age { get; set; }
    public char Gender { get; set; }
    public string Email { get; set; }
    public long Id { get; set; }
    public string Citizenship { get; set; }
    public string PhoneNumber { get; set; }
    public DateTime BirthDate { get; set; }
    public string Address { get; set; }
    public List<string> Languages { get; set; }

    // СОБЫТИЯ
    public event PersonEventHandler PersonCreated;
    public event PersonEventHandler ContactUpdated;
    public event PersonEventHandler InfoDisplayed;

    public Person(string name, int age, char gender, string email, long id, string citizenship, 
                 string phoneNumber, DateTime birthDate, string address, List<string> languages)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
        Id = id;
        Citizenship = citizenship;
        PhoneNumber = phoneNumber;
        BirthDate = birthDate;
        Address = address;
        Languages = languages;

        // Вызов события при создании
        PersonCreated?.Invoke($"Создан новый человек: {Name}");
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}, " +
                         $"Идентификационный номер: {Id}, Гражданство: {Citizenship}");
        InfoDisplayed?.Invoke($"Отображена информация о {Name}");
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}!");
    }

    public virtual void IsAdult()
    {
        Console.WriteLine($"Совершеннолетний(ая): {(Age > 18 ? "Да" : "Нет")}");
    }

    public virtual void DisplayContactInfo()
    {
        Console.WriteLine($"Контактная информация: Телефон: {PhoneNumber}, Email: {Email}, Адрес: {Address}");
    }

    public virtual void DisplayLanguages()
    {
        Console.Write($"Языки, которыми владеет {Name}: ");
        Console.WriteLine(string.Join(", ", Languages));
    }

    public void UpdateContactInfo(string newEmail, string newPhone, string newAddress)
    {
        Email = newEmail;
        PhoneNumber = newPhone;
        Address = newAddress;
        Console.WriteLine($"Полная контактная информация обновлена");
        ContactUpdated?.Invoke($"Контакты {Name} обновлены");
    }

    // Явная реализация интерфейса IPersonalInfo
    void IPersonalInfo.DisplayBasicInfo()
    {
        Console.WriteLine($"Базовая информация через интерфейс: {Name}, {Age} лет, ID: {Id}");
    }

    void IPersonalInfo.UpdateContact(string email, string phone)
    {
        this.Email = email;
        this.PhoneNumber = phone;
        Console.WriteLine($"Контакты обновлены через интерфейс: Email - {email}, Телефон - {phone}");
        ContactUpdated?.Invoke($"Контакты {Name} обновлены через интерфейс");
    }
}

// КЛАСС Student С КОЛЛЕКЦИЯМИ И СОБЫТИЯМИ
public class Student : Person, IPersonalInfo, IProfessional
{
    public string University { get; set; }
    public int Course { get; set; }
    public string Major { get; set; }
    public float Scholarship { get; set; }
    public double AverageGrade { get; set; }
    public List<string> CurrentSubjects { get; set; }
    public string StudentId { get; set; }
    public bool IsOnExchange { get; set; }

    // КОЛЛЕКЦИИ
    public Dictionary<string, double> Grades { get; set; }
    public HashSet<string> CompletedCourses { get; set; }
    public Queue<string> AssignmentQueue { get; set; }

    // СОБЫТИЯ
    public event WorkEventHandler AssignmentCompleted;
    public event PersonEventHandler GradeAdded;

    public Student(string name, int age, char gender, string email, long id, string citizenship,
                  string phoneNumber, DateTime birthDate, string address, List<string> languages,
                  string university, int course, string major, float scholarship, 
                  double averageGrade, List<string> currentSubjects, string studentId, bool isOnExchange)
        : base(name, age, gender, email, id, citizenship, phoneNumber, birthDate, address, languages)
    {
        University = university;
        Course = course;
        Major = major;
        Scholarship = scholarship;
        AverageGrade = averageGrade;
        CurrentSubjects = currentSubjects;
        StudentId = studentId;
        IsOnExchange = isOnExchange;

        // Инициализация коллекций
        Grades = new Dictionary<string, double>();
        CompletedCourses = new HashSet<string>();
        AssignmentQueue = new Queue<string>();
    }

    // МЕТОДЫ ДЛЯ РАБОТЫ С КОЛЛЕКЦИЯМИ
    public void AddGrade(string subject, double grade)
    {
        Grades[subject] = grade;
        GradeAdded?.Invoke($"Добавлена оценка {grade} по предмету {subject}");
        UpdateAverageGrade();
    }

    public void CompleteCourse(string courseName)
    {
        CompletedCourses.Add(courseName);
        Console.WriteLine($"Курс {courseName} завершен");
    }

    public void AddAssignment(string assignment)
    {
        AssignmentQueue.Enqueue(assignment);
        Console.WriteLine($"Добавлено задание: {assignment}");
    }

    public void CompleteAssignment()
    {
        if (AssignmentQueue.Count > 0)
        {
            var assignment = AssignmentQueue.Dequeue();
            Console.WriteLine($"Задание выполнено: {assignment}");
            AssignmentCompleted?.Invoke(assignment, 2); // 2 часа на выполнение
        }
    }

    private void UpdateAverageGrade()
    {
        if (Grades.Count > 0)
        {
            AverageGrade = Grades.Values.Average();
        }
    }

    public void DisplayGrades()
    {
        Console.WriteLine($"\nОценки студента {Name}:");
        foreach (var grade in Grades)
        {
            Console.WriteLine($"  {grade.Key}: {grade.Value}");
        }
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}, я студент {Course} курса университета «{University}».");
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Университет: {University}, Курс: {Course}, Специальность: {Major}, " +
                         $"Стипендия: {Scholarship} руб., Средний балл: {AverageGrade}");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} усердно учится по специальности {Major}");
    }

    public void TakeExam(string subject, string professor)
    {
        Console.WriteLine($"{Name} сдает экзамен по предмету: {subject} у профессора {professor}");
    }

    public void DisplayAcademicInfo()
    {
        Console.WriteLine($"Академическая информация: ID: {StudentId}, Средний балл: {AverageGrade}, " +
                         $"Программа обмена: {(IsOnExchange ? "Да" : "Нет")}");
        Console.WriteLine($"Текущие предметы: {string.Join(", ", CurrentSubjects)}");
    }

    // Явная реализация интерфейса IProfessional
    void IProfessional.DisplayProfessionalDetails()
    {
        Console.WriteLine($"Профессиональные данные студента: {Major} в {University}");
        Console.WriteLine($"Курс: {Course}, Стипендия: {Scholarship} руб.");
    }

    void IProfessional.PerformWork()
    {
        Study();
        Console.WriteLine($"Статус программы обмена: {(IsOnExchange ? "Участвует" : "Не участвует")}");
    }
}

// КЛАСС Employee С КОЛЛЕКЦИЯМИ И СОБЫТИЯМИ
public class Employee : Person, IPersonalInfo, IProfessional
{
    public string Company { get; set; }
    public int Salary { get; set; }
    public string JobTitle { get; set; }
    public string Department { get; set; }
    public int YearsOfService { get; set; }
    public string EmployeeId { get; set; }
    public DateTime HireDate { get; set; }
    public int VacationDays { get; set; }
    public string WorkSchedule { get; set; }

    // КОЛЛЕКЦИИ
    public Dictionary<string, int> Projects { get; set; } // проект -> часов затрачено
    public List<string> Skills { get; set; }
    public Stack<string> RecentTasks { get; set; }

    // СОБЫТИЯ
    public event WorkEventHandler TaskCompleted;
    public event PersonEventHandler ProjectStarted;

    public Employee(string name, int age, char gender, string email, long id, string citizenship,
                   string phoneNumber, DateTime birthDate, string address, List<string> languages,
                   string company, int salary, string jobTitle, string department, int yearsOfService,
                   string employeeId, DateTime hireDate, int vacationDays, string workSchedule)
        : base(name, age, gender, email, id, citizenship, phoneNumber, birthDate, address, languages)
    {
        Company = company;
        Salary = salary;
        JobTitle = jobTitle;
        Department = department;
        YearsOfService = yearsOfService;
        EmployeeId = employeeId;
        HireDate = hireDate;
        VacationDays = vacationDays;
        WorkSchedule = workSchedule;

        // Инициализация коллекций
        Projects = new Dictionary<string, int>();
        Skills = new List<string>();
        RecentTasks = new Stack<string>();
    }

    // МЕТОДЫ ДЛЯ РАБОТЫ С КОЛЛЕКЦИЯМИ
    public void AddProject(string projectName, int estimatedHours)
    {
        Projects[projectName] = estimatedHours;
        ProjectStarted?.Invoke($"Начат новый проект: {projectName}");
    }

    public void CompleteTask(string taskName, int hoursSpent)
    {
        RecentTasks.Push(taskName);
        Console.WriteLine($"Задача завершена: {taskName}");
        TaskCompleted?.Invoke(taskName, hoursSpent);
    }

    public void AddSkill(string skill)
    {
        if (!Skills.Contains(skill))
        {
            Skills.Add(skill);
            Console.WriteLine($"Добавлен навык: {skill}");
        }
    }

    public void DisplayRecentTasks()
    {
        Console.WriteLine($"\nПоследние задачи {Name}:");
        foreach (var task in RecentTasks)
        {
            Console.WriteLine($"  - {task}");
        }
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Компания: {Company}, Зарплата: {Salary} руб., Должность: {JobTitle}, " +
                         $"Отдел: {Department}, Стаж работы: {YearsOfService} лет");
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает на должности {JobTitle} в отделе {Department}");
    }

    public void TakeVacation(int days, string reason)
    {
        if (days <= VacationDays)
        {
            VacationDays -= days;
            Console.WriteLine($"{Name} взял отпуск на {days} дней по причине: {reason}. " +
                             $"Осталось дней отпуска: {VacationDays}");
        }
        else
        {
            Console.WriteLine($"Недостаточно дней отпуска. Доступно: {VacationDays}, Запрошено: {days}");
        }
    }

    public int CalculateBonus(int customBonusFactor)
    {
        return YearsOfService * customBonusFactor;
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА IProfessional
    void IProfessional.DisplayProfessionalDetails()
    {
        Console.WriteLine($"Профессиональные данные сотрудника: {JobTitle} в {Company}");
        Console.WriteLine($"Зарплата: {Salary} руб., Стаж: {YearsOfService} лет, Отдел: {Department}");
    }

    void IProfessional.PerformWork()
    {
        Work();
        Console.WriteLine($"График работы: {WorkSchedule}, Отпускных дней осталось: {VacationDays}");
    }
}

// КЛАСС Teacher С КОЛЛЕКЦИЯМИ И СОБЫТИЯМИ
public class Teacher : Person, IPersonalInfo, IProfessional
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string Qualification { get; set; }
    public bool PhDDegree { get; set; }
    public int HoursPerWeek { get; set; }
    public string TeacherId { get; set; }
    public List<string> ResearchInterests { get; set; }
    public int PublicationsCount { get; set; }
    public string OfficeNumber { get; set; }

    // КОЛЛЕКЦИИ
    public Dictionary<string, List<string>> CourseMaterials { get; set; } // курс -> материалы
    public HashSet<string> TeachingCertifications { get; set; }
    public Queue<string> StudentAppointments { get; set; }

    // СОБЫТИЯ
    public event WorkEventHandler LectureConducted;
    public event PersonEventHandler StudentGraded;

    public Teacher(string name, int age, char gender, string email, long id, string citizenship,
                  string phoneNumber, DateTime birthDate, string address, List<string> languages,
                  string subject, int experience, string qualification, bool phdDegree, int hoursPerWeek,
                  string teacherId, List<string> researchInterests, int publicationsCount, string officeNumber)
        : base(name, age, gender, email, id, citizenship, phoneNumber, birthDate, address, languages)
    {
        Subject = subject;
        Experience = experience;
        Qualification = qualification;
        PhDDegree = phdDegree;
        HoursPerWeek = hoursPerWeek;
        TeacherId = teacherId;
        ResearchInterests = researchInterests;
        PublicationsCount = publicationsCount;
        OfficeNumber = officeNumber;

        // Инициализация коллекций
        CourseMaterials = new Dictionary<string, List<string>>();
        TeachingCertifications = new HashSet<string>();
        StudentAppointments = new Queue<string>();
    }

    // МЕТОДЫ ДЛЯ РАБОТЫ С КОЛЛЕКЦИЯМИ
    public void AddCourseMaterial(string course, string material)
    {
        if (!CourseMaterials.ContainsKey(course))
        {
            CourseMaterials[course] = new List<string>();
        }
        CourseMaterials[course].Add(material);
        Console.WriteLine($"Добавлен материал для курса {course}: {material}");
    }

    public void AddCertification(string certification)
    {
        TeachingCertifications.Add(certification);
        Console.WriteLine($"Добавлена сертификация: {certification}");
    }

    public void ScheduleAppointment(string studentName)
    {
        StudentAppointments.Enqueue(studentName);
        Console.WriteLine($"Запланирована встреча со студентом: {studentName}");
    }

    public void ConductNextAppointment()
    {
        if (StudentAppointments.Count > 0)
        {
            var student = StudentAppointments.Dequeue();
            Console.WriteLine($"Проведена встреча со студентом: {student}");
        }
    }

    public void DisplayCertifications()
    {
        Console.WriteLine($"\nСертификации преподавателя {Name}:");
        foreach (var cert in TeachingCertifications)
        {
            Console.WriteLine($"  - {cert}");
        }
    }

    // ВСЕ СУЩЕСТВУЮЩИЕ МЕТОДЫ ОСТАЮТСЯ
    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}, я преподаватель дисциплины «{Subject}» " +
                         $"с опытом работы - {Experience} года.");
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Предмет: {Subject}, Опыт работы: {Experience} лет, Квалификация: {Qualification}, " +
                         $"Докторская степень: {(PhDDegree ? "Да" : "Нет")}, Нагрузка: {HoursPerWeek} часов в неделю");
    }

    public void ConductLecture()
    {
        Console.WriteLine($"{Name} проводит лекцию по предмету {Subject}");
        LectureConducted?.Invoke($"Лекция по {Subject}", 2);
    }

    public void GradeStudents(string subject, int studentCount, string assignmentType)
    {
        Console.WriteLine($"{Name} проверяет {assignmentType} {studentCount} студентов по предмету {subject}");
        StudentGraded?.Invoke($"Оценены {studentCount} студентов по {subject}");
    }

    public void DisplayResearchInfo()
    {
        Console.WriteLine($"Научная деятельность: Публикации: {PublicationsCount}, " +
                         $"Области исследований: {string.Join(", ", ResearchInterests)}");
    }

    public decimal CalculateAnnualIncome(decimal baseSalaryPerHour, decimal publicationBonusRate)
    {
        decimal annualBaseIncome = HoursPerWeek * 52 * baseSalaryPerHour;
        decimal degreeBonus = PhDDegree ? 100_000m : 0;
        decimal publicationBonus = PublicationsCount * publicationBonusRate;
        return annualBaseIncome + degreeBonus + publicationBonus;
    }

    // Явная реализация интерфейса IProfessional
    void IProfessional.DisplayProfessionalDetails()
    {
        Console.WriteLine($"Профессиональные данные преподавателя: {Subject}, Опыт: {Experience} лет");
        Console.WriteLine($"Квалификация: {Qualification}, Докторская степень: {(PhDDegree ? "Да" : "Нет")}");
    }

    void IProfessional.PerformWork()
    {
        ConductLecture();
        Console.WriteLine($"Нагрузка: {HoursPerWeek} часов/неделю, Кабинет: {OfficeNumber}");
    }
}

// КЛАСС Intern С КОЛЛЕКЦИЯМИ И СОБЫТИЯМИ
public class Intern : Employee, IPersonalInfo, IProfessional
{
    public string Mentor { get; set; }
    public DateTime StartDate { get; set; }
    public DateTime EndDate { get; set; }
    public string InternshipType { get; set; }
    public List<string> LearningGoals { get; set; }
    public bool IsPaid { get; set; }
    public string FinalProject { get; set; }

    // КОЛЛЕКЦИИ
    public Dictionary<string, string> SkillAssessments { get; set; } // навык -> оценка
    public List<string> CompletedTrainings { get; set; }
    public Stack<string> WeeklyReports { get; set; }

    // СОБЫТИЯ
    public event PersonEventHandler TrainingCompleted;
    public event WorkEventHandler ProjectMilestoneReached;

    public Intern(string name, int age, char gender, string email, long id, string citizenship,
                 string phoneNumber, DateTime birthDate, string address, List<string> languages,
                 string company, int salary, string jobTitle, string department, int yearsOfService,
                 string employeeId, DateTime hireDate, int vacationDays, string workSchedule,
                 string mentor, DateTime startDate, DateTime endDate, string internshipType,
                 List<string> learningGoals, bool isPaid, string finalProject)
        : base(name, age, gender, email, id, citizenship, phoneNumber, birthDate, address, languages,
               company, salary, jobTitle, department, yearsOfService, employeeId, hireDate, vacationDays, workSchedule)
    {
        Mentor = mentor;
        StartDate = startDate;
        EndDate = endDate;
        InternshipType = internshipType;
        LearningGoals = learningGoals;
        IsPaid = isPaid;
        FinalProject = finalProject;

        // Инициализация коллекций
        SkillAssessments = new Dictionary<string, string>();
        CompletedTrainings = new List<string>();
        WeeklyReports = new Stack<string>();
    }

    // МЕТОДЫ ДЛЯ РАБОТЫ С КОЛЛЕКЦИЯМИ
    public void AddSkillAssessment(string skill, string assessment)
    {
        SkillAssessments[skill] = assessment;
        Console.WriteLine($"Оценка навыка {skill}: {assessment}");
    }

    public void CompleteTraining(string training)
    {
        CompletedTrainings.Add(training);
        TrainingCompleted?.Invoke($"Завершено обучение: {training}");
    }

    public void AddWeeklyReport(string report)
    {
        WeeklyReports.Push(report);
        Console.WriteLine($"Добавлен еженедельный отчет: {report}");
    }

    public void ReachProjectMilestone(string milestone, int hoursSpent)
    {
        Console.WriteLine($"Достигнут этап проекта: {milestone}");
        ProjectMilestoneReached?.Invoke(milestone, hoursSpent);
    }

    public void DisplaySkillAssessments()
    {
        Console.WriteLine($"\nОценки навыков стажера {Name}:");
        foreach (var assessment in SkillAssessments)
        {
            Console.WriteLine($"  {assessment.Key}: {assessment.Value}");
        }
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Наставник: {Mentor}, Начало стажировки: {StartDate.ToShortDateString()}, " +
                         $"Конец стажировки: {EndDate.ToShortDateString()}, Тип стажировки: {InternshipType}");
    }

    public void AttendTraining()
    {
        Console.WriteLine($"{Name} посещает учебные занятия в рамках стажировки");
    }

    public void WorkOnProject()
    {
        Console.WriteLine($"{Name} работает над финальным проектом: {FinalProject}");
    }

    public void DisplayInternshipGoals()
    {
        Console.WriteLine($"Цели обучения стажера {Name}:");
        foreach (var goal in LearningGoals)
        {
            Console.WriteLine($"- {goal}");
        }
    }

    public TimeSpan RemainingInternshipDuration()
    {
        return EndDate - DateTime.Now;
    }

    // Явная реализация IProfessional
    void IProfessional.DisplayProfessionalDetails()
    {
        Console.WriteLine($"Профессиональные данные стажера: {JobTitle} в {Company}");
        Console.WriteLine($"Наставник: {Mentor}, Тип стажировки: {InternshipType}, Оплачиваемая: {(IsPaid ? "Да" : "Нет")}");
    }

    void IProfessional.PerformWork()
    {
        WorkOnProject();
        AttendTraining();
        Console.WriteLine($"Осталось дней стажировки: {RemainingInternshipDuration().Days}");
    }
}

// GENERIC МЕНЕДЖЕР С КОЛЛЕКЦИЯМИ И СОБЫТИЯМИ
public class PersonManager<T> where T : Person
{
    private List<T> _people;
    private Dictionary<long, T> _peopleById;

    // СОБЫТИЕ
    public event Action<string, T> PersonAdded;
    public event Action<string, T> PersonRemoved;

    public PersonManager()
    {
        _people = new List<T>();
        _peopleById = new Dictionary<long, T>();
    }

    public void AddPerson(T person)
    {
        _people.Add(person);
        _peopleById[person.Id] = person;
        PersonAdded?.Invoke($"Добавлен новый человек в менеджер", person);
    }

    public void RemovePerson(T person)
    {
        if (_people.Remove(person))
        {
            _peopleById.Remove(person.Id);
            PersonRemoved?.Invoke($"Удален человек из менеджера", person);
        }
    }

    public T FindPersonById(long id)
    {
        return _peopleById.ContainsKey(id) ? _peopleById[id] : null;
    }

    public void DisplayAllInfo()
    {
        Console.WriteLine($"\n=== Информация о всех {typeof(T).Name} ===");
        foreach (var person in _people)
        {
            person.GetInfo();
            Console.WriteLine("---");
        }
    }

    public void DisplayAllContactInfo()
    {
        Console.WriteLine($"\n=== Контактная информация всех {typeof(T).Name} ===");
        foreach (var person in _people)
        {
            person.DisplayContactInfo();
        }
    }

    public List<T> GetPeopleByCondition(Func<T, bool> condition)
    {
        return _people.Where(condition).ToList();
    }

    // РАБОТА С РАЗЛИЧНЫМИ КОЛЛЕКЦИЯМИ
    public HashSet<string> GetAllUniqueNames()
    {
        return new HashSet<string>(_people.Select(p => p.Name));
    }

    public Dictionary<string, List<T>> GroupByCitizenship()
    {
        return _people.GroupBy(p => p.Citizenship)
                     .ToDictionary(g => g.Key, g => g.ToList());
    }

    public Queue<T> GetPeopleQueue()
    {
        return new Queue<T>(_people);
    }

    public Stack<T> GetPeopleStack()
    {
        return new Stack<T>(_people);
    }
}

// ОБРАБОТЧИКИ СОБЫТИЙ
void OnPersonCreated(string message) => Console.WriteLine($"[СОБЫТИЕ] {message}");
void OnContactUpdated(string message) => Console.WriteLine($"[СОБЫТИЕ] {message}");
void OnAssignmentCompleted(string task, int hours) => 
    Console.WriteLine($"[СОБЫТИЕ] Задание '{task}' выполнено за {hours} часов");
void OnTaskCompleted(string task, int hours) => 
    Console.WriteLine($"[СОБЫТИЕ] Задача '{task}' завершена за {hours} часов");
void OnLectureConducted(string lecture, int hours) => 
    Console.WriteLine($"[СОБЫТИЕ] Лекция '{lecture}' проведена за {hours} часов");

// СОЗДАНИЕ ОБЪЕКТОВ
var person = new Person("Иван Иванов", 30, 'М', "ivan@example.com", 1, "РФ",
                        "+7-999-123-45-67", new DateTime(1993, 5, 15), "ул. Пушкина, д.10",
                        new List<string> { "Русский", "Английский" });

var student = new Student("Михаил Соколов", 20, 'М', "mikhail@example.com", 2, "РФ",
                        "+7-999-123-45-68", new DateTime(2003, 8, 20), "ул. Лермонтова, д.15",
                        new List<string> { "Русский", "Английский", "Немецкий" },
                        "МФТИ", 3, "Компьютерные науки", 10000, 4.5,
                        new List<string> { "Математика", "Программирование", "Физика" },
                        "STU001", false);

var employee = new Employee("Анна Петрова", 35, 'Ж', "anna@example.com", 3, "РФ",
                        "+7-999-123-45-69", new DateTime(1988, 3, 10), "ул. Толстого, д.20",
                        new List<string> { "Русский", "Французский" },
                        "Яндекс", 120000, "Менеджер проектов", "IT-отдел", 10,
                        "EMP001", new DateTime(2013, 6, 1), 28, "5/2");

var teacher = new Teacher("Владимир Кузнецов", 45, 'М', "vladimir@example.com", 4, "РФ",
                        "+7-999-123-45-70", new DateTime(1978, 12, 5), "ул. Чехова, д.25",
                        new List<string> { "Русский", "Английский", "Испанский" },
                        "Математика", 20, "Высшая категория", true, 18,
                        "TCH001", new List<string> { "Алгебра", "Теория чисел" }, 15, "А-101");

// ПОДПИСКА НА СОБЫТИЯ
student.PersonCreated += OnPersonCreated;
student.ContactUpdated += OnContactUpdated;
student.AssignmentCompleted += OnAssignmentCompleted;
student.GradeAdded += OnPersonCreated;

employee.PersonCreated += OnPersonCreated;
employee.TaskCompleted += OnTaskCompleted;
employee.ProjectStarted += OnPersonCreated;

teacher.PersonCreated += OnPersonCreated;
teacher.LectureConducted += OnLectureConducted;
teacher.StudentGraded += OnPersonCreated;

Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");
Person[] people = { person, student, employee, teacher };

foreach (var p in people)
{
    p.SayHello();
    p.GetInfo();
    Console.WriteLine();
}

Console.WriteLine("=== ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ ===");
person.UpdateContactInfo("newivan3@example.com", "+7-888-888-88-88", "новый адрес");
Console.WriteLine();

student.TakeExam("Физика", "Профессор Сидоров");
Console.WriteLine();

employee.TakeVacation(7, "семейные обстоятельства");
Console.WriteLine();

teacher.GradeStudents("Алгебра", 25, "курсовые работы");
Console.WriteLine();

// РАБОТА С КОЛЛЕКЦИЯМИ
Console.WriteLine("=== РАБОТА С КОЛЛЕКЦИЯМИ СТУДЕНТА ===");
student.AddGrade("Математика", 4.5);
student.AddGrade("Программирование", 5.0);
student.AddGrade("Физика", 4.0);
student.DisplayGrades();

student.AddAssignment("Лабораторная работа по программированию");
student.AddAssignment("Курсовая по математике");
student.CompleteAssignment();
student.CompleteAssignment();

// РАБОТА С КОЛЛЕКЦИЯМИ СОТРУДНИКА
Console.WriteLine("\n=== РАБОТА С КОЛЛЕКЦИЯМИ СОТРУДНИКА ===");
employee.AddProject("Разработка новой CRM", 200);
employee.AddProject("Миграция на новую БД", 150);

employee.AddSkill("C#");
employee.AddSkill("SQL");
employee.AddSkill("Project Management");

employee.CompleteTask("Проектирование архитектуры", 8);
employee.CompleteTask("Написание документации", 4);
employee.DisplayRecentTasks();

// РАБОТА С КОЛЛЕКЦИЯМИ ПРЕПОДАВАТЕЛЯ
Console.WriteLine("\n=== РАБОТА С КОЛЛЕКЦИЯМИ ПРЕПОДАВАТЕЛЯ ===");
teacher.AddCourseMaterial("Алгебра", "Лекция 1: Основные понятия");
teacher.AddCourseMaterial("Алгебра", "Практическое задание 1");
teacher.AddCertification("Методика преподавания");
teacher.AddCertification("Цифровые образовательные технологии");
teacher.DisplayCertifications();

teacher.ScheduleAppointment("Иван Петров");
teacher.ScheduleAppointment("Мария Сидорова");
teacher.ConductNextAppointment();

// РАБОТА С GENERIC МЕНЕДЖЕРОМ
Console.WriteLine("\n=== РАБОТА С GENERIC МЕНЕДЖЕРОМ ===");
var studentManager = new PersonManager<Student>();

// Подписка на события менеджера
studentManager.PersonAdded += (message, person) => 
    Console.WriteLine($"[МЕНЕДЖЕР] {message}: {person.Name}");
studentManager.PersonRemoved += (message, person) => 
    Console.WriteLine($"[МЕНЕДЖЕР] {message}: {person.Name}");

studentManager.AddPerson(student);

var anotherStudent = new Student("Елена Волкова", 21, 'Ж', "elena@example.com", 5, "РФ",
                                "+7-999-123-45-71", new DateTime(2002, 7, 12), "ул. Есенина, д.30",
                                new List<string> { "Русский", "Английский" },
                                "МГУ", 4, "Физика", 12000, 4.7,
                                new List<string> { "Квантовая физика", "Термодинамика" },
                                "STU002", true);

studentManager.AddPerson(anotherStudent);
studentManager.DisplayAllInfo();

// ИСПОЛЬЗОВАНИЕ ЛЯМБДА-ВЫРАЖЕНИЙ ДЛЯ ФИЛЬТРАЦИИ
var topStudents = studentManager.GetPeopleByCondition(s => s.AverageGrade > 4.6);
Console.WriteLine("\n=== Студенты со средним баллом выше 4.6 ===");
foreach (var topStudent in topStudents)
{
    topStudent.GetInfo();
}

// РАБОТА С РАЗЛИЧНЫМИ КОЛЛЕКЦИЯМИ В МЕНЕДЖЕРЕ
var uniqueNames = studentManager.GetAllUniqueNames();
Console.WriteLine($"\nУникальные имена: {string.Join(", ", uniqueNames)}");

// ДЕМОНСТРАЦИЯ ДЕЛЕГАТОВ
Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ДЕЛЕГАТОВ ===");
var personalService = new PersonalInfoService(student);

// Добавляем обработчики к делегатам
personalService.OnInfoDisplay += () => Console.WriteLine(">>> Готовим данные...");
personalService.OnContactUpdate += (msg) => Console.WriteLine($">>> Обновление: {msg}");

personalService.ProcessPersonalInfo();
personalService.UpdateContacts("new_email@example.com", "+7-999-000-00-00");

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ НОВЫХ МЕТОДОВ ===");
student.DisplayAcademicInfo();
teacher.DisplayResearchInfo();

Console.WriteLine($"Годовая зарплата с повышенным бонусом за публикации: {teacher.CalculateAnnualIncome(1000, 7000)} руб.");

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ИНТЕРФЕЙСОВ И УПРАВЛЕНИЯ ЗАВИСИМОСТЯМИ ===");

// Работа с сервисами
Console.WriteLine("\n=== РАБОТА С СЕРВИСАМИ ===");
var personalService2 = new PersonalInfoService(student);
personalService2.ProcessPersonalInfo();

var professionalService = new ProfessionalService(student);
professionalService.ProcessProfessionalInfo();

Console.WriteLine();

var employeeProfessionalService = new ProfessionalService(employee);
employeeProfessionalService.ProcessProfessionalInfo();

Console.WriteLine();

var teacherProfessionalService = new ProfessionalService(teacher);
teacherProfessionalService.ProcessProfessionalInfo();

// Работа с явной реализацией интерфейсов
Console.WriteLine("\n=== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ ===");

IPersonalInfo personalInfo = student;
personalInfo.DisplayBasicInfo();
personalInfo.UpdateContact("new_student@example.com", "+7-999-000-00-00");

Console.WriteLine();

IProfessional professional = teacher;
professional.DisplayProfessionalDetails();
professional.PerformWork();

Console.WriteLine();

// Демонстрация полиморфизма через интерфейсы
Console.WriteLine("=== ПОЛИМОРФИЗМ ЧЕРЕЗ ИНТЕРФЕЙСЫ ===");
IProfessional[] professionals = { student, employee, teacher };
foreach (var prof in professionals)
{
    prof.DisplayProfessionalDetails();
    Console.WriteLine("---");
}

// ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ КОЛЛЕКЦИЙ
Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНАЯ ДЕМОНСТРАЦИЯ КОЛЛЕКЦИЙ ===");

// Queue для студента
student.AddAssignment("Домашнее задание по физике");
student.AddAssignment("Подготовка к экзамену");

// Stack для сотрудника
employee.CompleteTask("Тестирование системы", 6);
employee.CompleteTask("Code Review", 2);

// Dictionary для группировки
var employeeManager = new PersonManager<Employee>();
employeeManager.AddPerson(employee);
var groupedByCitizenship = employeeManager.GroupByCitizenship();

Console.WriteLine("Группировка по гражданству:");
foreach (var group in groupedByCitizenship)
{
    Console.WriteLine($"  {group.Key}: {group.Value.Count} человек");
}

// HashSet для уникальных значений
teacher.AddCertification("Методика преподавания"); // Дубликат - не добавится
teacher.AddCertification("Новая сертификация"); // Добавится
```
