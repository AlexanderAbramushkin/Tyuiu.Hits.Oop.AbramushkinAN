```polyglot-notebook
using System;
using System.Collections.Generic;

public abstract class Component
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public decimal Price { get; set; }
    public string Model { get; set; }

    public Component(string name, string manufacturer, string model, decimal price)
    {
        Name = name;
        Manufacturer = manufacturer;
        Model = model;
        Price = price;
    }

    // Виртуальный метод GetInfo
    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Модель: {Model}");
        Console.WriteLine($"Цена: {Price} руб.");
    }

    // Перегрузка оператора ToString
    public override string ToString()
    {
        return $"{Manufacturer} {Name} {Model} - {Price} руб.";
    }
}

// Интерфейс Installable
public interface Installable
{
    void Install();
}

// Класс CPU
public class CPU : Component, Installable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string name, string manufacturer, string model, decimal price, 
               int cores, double frequency) 
               : base(name, manufacturer, model, price)
    {
        Cores = cores;
        Frequency = frequency;
    }

    // Перекрытие метода GetInfo
    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Ядра: {Cores}");
        Console.WriteLine($"Частота: {Frequency} GHz");
        Console.WriteLine("---");
    }

    // Реализация интерфейса Installable
    public void Install()
    {
        Console.WriteLine($"Устанавливаю процессор {Manufacturer} {Model}");
        Console.WriteLine("Закрепляю кулер...");
        Console.WriteLine("Процессор установлен!");
    }

    public override string ToString()
    {
        return base.ToString() + $" [{Cores} ядер, {Frequency} GHz]";
    }
}

// Класс RAM
public class RAM : Component, Installable
{
    public int Capacity { get; set; }
    public double Frequency { get; set; }

    public RAM(string name, string manufacturer, string model, decimal price,
               int capacity, double frequency)
               : base(name, manufacturer, model, price)
    {
        Capacity = capacity;
        Frequency = frequency;
    }

    // Перекрытие метода GetInfo
    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объем: {Capacity} GB");
        Console.WriteLine($"Частота: {Frequency} MHz");
        Console.WriteLine("---");
    }

    // Реализация интерфейса Installable
    public void Install()
    {
        Console.WriteLine($"Устанавливаю оперативную память {Manufacturer} {Model}");
        Console.WriteLine("Вставляю модуль памяти до щелчка...");
        Console.WriteLine("Оперативная память установлена!");
    }

    public override string ToString()
    {
        return base.ToString() + $" [{Capacity} GB]";
    }
}

// Класс GPU
public class GPU : Component, Installable
{
    public int VRAM { get; set; }
    public string Interface { get; set; }

    public GPU(string name, string manufacturer, string model, decimal price,
               int vram, string interfaceType)
               : base(name, manufacturer, model, price)
    {
        VRAM = vram;
        Interface = interfaceType;
    }

    // Перекрытие метода GetInfo
    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Видеопамять: {VRAM} GB");
        Console.WriteLine($"Интерфейс: {Interface}");
        Console.WriteLine("---");
    }

    // Реализация интерфейса Installable
    public void Install()
    {
        Console.WriteLine($"Устанавливаю видеокарту {Manufacturer} {Model}");
        Console.WriteLine($"Вставляю в слот {Interface}...");
        Console.WriteLine("Закрепляю винтом...");
        Console.WriteLine("Видеокарта установлена!");
    }

    public override string ToString()
    {
        return base.ToString() + $" [{VRAM} GB VRAM]";
    }
}

// Класс HDD
public class HDD : Component, Installable
{
    public int Capacity { get; set; }
    public string Interface { get; set; }
    public int RPM { get; set; }

    public HDD(string name, string manufacturer, string model, decimal price,
               int capacity, string interfaceType, int rpm)
               : base(name, manufacturer, model, price)
    {
        Capacity = capacity;
        Interface = interfaceType;
        RPM = rpm;
    }

    // Перекрытие метода GetInfo
    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Объем: {Capacity} GB");
        Console.WriteLine($"Интерфейс: {Interface}");
        Console.WriteLine($"Скорость вращения: {RPM} RPM");
        Console.WriteLine("---");
    }

    // Реализация интерфейса Installable
    public void Install()
    {
        Console.WriteLine($"Устанавливаю жесткий диск {Manufacturer} {Model}");
        Console.WriteLine("Закрепляю в отсеке...");
        Console.WriteLine($"Подключаю интерфейс {Interface}...");
        Console.WriteLine("Жесткий диск установлен!");
    }

    public override string ToString()
    {
        return base.ToString() + $" [{Capacity} GB, {RPM} RPM]";
    }
}

// Класс Computer
public class Computer
{
    private List<Component> components;

    public string Name { get; set; }
    public Computer(string name)
    {
        Name = name;
        components = new List<Component>();
    }

    // Generic метод для добавления компонента
    public void AddComponent<T>(T component) where T : Component, Installable
    {
        components.Add(component);
        Console.WriteLine($"Добавлен компонент: {component}");
        component.Install();
        Console.WriteLine();
    }

    // Generic метод для удаления компонента
    public bool RemoveComponent<T>(T component) where T : Component
    {
        bool removed = components.Remove(component);
        if (removed)
        {
            Console.WriteLine($"Удален компонент: {component}");
        }
        return removed;
    }

    // Метод для получения общей стоимости
    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }
}

public static class ComputerFactory
{
    // Метод для создания офисного компьютера
    public static Computer CreateOfficeComputer(string name)
    {
        var computer = new Computer(name);
        
        var cpu = new CPU("Процессор", "AMD", "Ryzen 5 5600G", 15000, 6, 3.9);
        var ram = new RAM("Оперативная память", "Kingston", "ValueRAM 16GB", 4000, 16, 2666);
        var hdd = new HDD("Жесткий диск", "Western Digital", "Blue 1TB", 3500, 1000, "SATA III", 7200);

        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(hdd);

        return computer;
    }
}


Console.WriteLine("=== ДЕМОНСТРАЦИЯ СИСТЕМЫ КОМПЬЮТЕРНЫХ КОМПОНЕНТОВ ===\n");

Console.WriteLine("1. ДЕМОНСТРАЦИЯ ВЫВОДА КОМПОНЕНТОВ:");
Console.WriteLine("--------------------------------------");

List<Component> components = new List<Component>
{
    new CPU("Процессор", "Intel", "Core i5-12400", 20000, 6, 2.5),
    new RAM("Оперативная память", "Crucial", "Ballistix 16GB", 4500, 16, 3000),
    new GPU("Видеокарта", "AMD", "RX 6700 XT", 40000, 12, "PCIe 4.0"),
    new HDD("Жесткий диск", "Toshiba", "P300 1TB", 4000, 1000, "SATA III", 7200)
};
foreach (var component in components)
{
    component.GetInfo();
}
Console.WriteLine("--------------------------------------");
Console.WriteLine();

Console.WriteLine("2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ ОПЕРАТОРОВ != и ==:");
Console.WriteLine("--------------------------------------");
var cpu1 = new CPU("Процессор", "Intel", "Core i5", 20000, 6, 2.5);
var cpu2 = new CPU("Процессор", "AMD", "Ryzen 5", 18000, 6, 3.2);

Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}"); // True

// Демонстрация перегрузки ToString
Console.WriteLine($"\nПерегрузка ToString(): {cpu1}");
Console.WriteLine("--------------------------------------");
Console.WriteLine();
Console.WriteLine("\n3. ДЕМОНСТРАЦИЯ ФАБРИКИ КОМПЬЮТЕРОВ:");
Console.WriteLine("--------------------------------------");
Console.WriteLine("Офисный ПК 'Стандарт'");
var officePC = ComputerFactory.CreateOfficeComputer("Офисный ПК 'Стандарт'");

Console.WriteLine("--------------------------------------");
Console.WriteLine();
Console.WriteLine("4. ДЕМОНСТРАЦИЯ GENERIC МЕТОДОВ:");
Console.WriteLine("--------------------------------------");
var customPC = new Computer("Кастомный ПК");

// Добавление компонентов с использованием Generic
var newCPU = new CPU("Процессор", "AMD", "Ryzen 7 5800X", 25000, 8, 3.8);
var newRAM = new RAM("Оперативная память", "G.Skill", "Trident Z 32GB", 9000, 32, 3600);

// Добавление компонентов
customPC.AddComponent(newCPU);
customPC.AddComponent(newRAM);

// Удаление компонента
customPC.RemoveComponent(newRAM);
Console.WriteLine("--------------------------------------");
Console.WriteLine();
Console.WriteLine("=== ДЕМОНСТРАЦИЯ ЗАВЕРШЕНА ===");
Console.ReadLine();
```
