```polyglot-notebook
public class Employee
{
    public string Name{get;set;}
    public string Position {get;set;}
    public double Salary {get;set;}

    public Employee(string name,string position, double salary) : base()
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}\nДолжность: {Position}\nЗарплата {Salary}");
    }
}

public class Manager : Employee
{
    public string Department{get;set;}

    public Manager(string name,string position, double salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

interface IReportable
{
    string GenerateReport();
}

interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName{get;set;}


    public string[] TeamMembers{get;set;}

    public ProjectManager(string name, string position, double salary, string department, string projectname, string[] team) 
        : base(name, position, salary, department)
    {
        ProjectName = projectname;
        TeamMembers = team;
    }

        public string GenerateReport()
    {
        return $"Отчет по проекту '{ProjectName}'. Кол-во человек в команде: {TeamMembers.Length} человек. Статус проекта: В работе";
    }

    public string[] GetTeamMembers()
    {
        return TeamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
        Console.WriteLine($"Размер команды: {TeamMembers.Length} человек");
    }
}

Employee myEmployee = new Employee("Егор Евссев", "AI-инженер", 70000);
Manager myManager = new Manager("Виктор Анищенко", "Разработчик", 100000, "AI-разработка");
string[] teamMembers = {"Арсений Иванов", "Иван Сидоров", "Мария Аханова"};
ProjectManager myProjectManager = new ProjectManager("Юлия Брезина", "Менеджер проектов", 120000, "Разработка мобильных приложений", "Чат-бот для психологов",teamMembers);

Console.WriteLine("=====Сотрудник=====");
myEmployee.PrintInfo();
Console.WriteLine();

Console.WriteLine("=====Менеджер=====");
myManager.PrintInfo();
Console.WriteLine();

Console.WriteLine("=====Проектный менеджер=====");
myProjectManager.PrintInfo();
Console.WriteLine();

string report = myProjectManager.GenerateReport();
Console.WriteLine($"Отчёт: {report}");
Console.WriteLine();

string[] team = myProjectManager.GetTeamMembers();
Console.WriteLine("Члены команды:");
foreach (string member in team)
{
    Console.WriteLine($"  - {member}");
}
```
