### Вариант задания №1


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма<br>
Требования к базовому классу Person:<br>
• Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).<br>
• Методы:<br>
o GetInfo(): метод для получения информации о человеке в виде строки.<br>
o SayHello(): метод для вывода приветствия от человека.<br>
Требования к производным классам:<br>
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как
Университет (University) и Курс (Course). Метод SayHello() должен быть
переопределен для добавления информации о курсе при приветствии.<br>
2. Работник (Employee): Должен содержать дополнительные атрибуты, такие
как Компания (subject) и Зарплата (Salary). Метод GetInfo() должен быть
переопределен для включения информации о компании и зарплате.<br>
3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы
(Experience). Метод SayHello() должен быть переопределен для добавления
информации о преподаваемом предмете при приветствии.<br>

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public char Gender { get; set; }
    public string Email { get; set; }
    public long Id { get; set; }
    public string Citizenship { get; set; } 

    public Person(string name, int age, char gender, string email, long id, string citizenship)
    {
        Name = name;
        Age = age;
        Gender = gender;
        Email = email;
        Id = id;
        Citizenship = citizenship;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine( $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Email: {Email}, Идентификационный номер: {Id}, Гражданство: {Citizenship}");
    }

    public virtual void SayHello()
    {
        Console.WriteLine( $"Привет, меня зовут {Name}!");
    }

    public virtual void IsAdult()
    {
        Console.WriteLine($"Совершеннолетний(ая): {(Age>18 ? "Да":"Нет")}");
    }
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }
    public string Major { get; set; }       
    public float Scholarship { get; set; }
    public Student(string name, int age, char gender, string email, long id, string citizenship, string university, int course, string major, float scholarship) : base(name, age, gender, email, id, citizenship)
    {
        University = university;
        Course = course;
        Major = major;
        Scholarship = scholarship;
    }

    public override void SayHello()
    {
        Console.WriteLine( $"Привет, меня зовут {Name}, я студент {Course} курса университета «{University}».");
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine( $"Университет: {University}, Курс: {Course}, Специальность: {Major}, Стипендия: {Scholarship} руб.");
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public int Salary { get; set; }
    public string JobTitle { get; set; }        
    public string Department { get; set; }        
    public int YearsOfService { get; set; }       

    public Employee(string name, int age, char gender, string email, long id, string citizenship, string company, int salary, string jobTitle, string department, int yearsOfService) : base(name, age, gender, email, id, citizenship)
    {
        Company = company;
        Salary = salary;
        JobTitle = jobTitle;
        Department = department;
        YearsOfService = yearsOfService;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine( $"Компания: {Company}, Зарплата: {Salary} руб., Должность: {JobTitle}, Отдел: {Department}, Стаж работы: {YearsOfService} лет");
    }


    public int CalculateBonus()
    {
        const int bonusFactor = 5_000; 
        return YearsOfService * bonusFactor;
    }
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string Qualification { get; set; } 
    public bool PhDDegree { get; set; }  
    public int HoursPerWeek { get; set; }

    public Teacher(string name, int age, char gender, string email, long id, string citizenship, string subject, int experience, string qualification, bool phdDegree, int hoursPerWeek) : base(name, age, gender, email, id, citizenship)
    {
        Subject = subject;
        Experience = experience;
        Qualification = qualification;
        PhDDegree = phdDegree;
        HoursPerWeek = hoursPerWeek;
    }

    public override void SayHello()
    {
        Console.WriteLine( $"Привет, меня зовут {Name}, я преподаватель дисциплины «{Subject}» с опытом работы - {Experience} года.");
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine( $"Предмет: {Subject}, Опыт работы: {Experience} лет, Квалификация: {Qualification}, Докторская степень: {(PhDDegree ? "Да" : "Нет")}, Нагрузка: {HoursPerWeek} часов в неделю");
    }

    public decimal CalculateAnnualIncome(decimal baseSalaryPerHour)
    {
        decimal annualBaseIncome = HoursPerWeek * 52 * baseSalaryPerHour;
        decimal degreeBonus = PhDDegree ? 100_000m : 0;        
        return annualBaseIncome + degreeBonus;
    }
}

public class Intern : Employee
{
    public string Mentor { get; set; }
    public DateTime StartDate { get; set; }
    public DateTime EndDate{get;set;}    

    public Intern(string name, int age, char gender, string email, long id, string citizenship, string company, int salary, string jobTitle, string department, int yearsOfService, string mentor, DateTime startDate, DateTime endDate) : base(name, age, gender, email, id, citizenship, company, salary, jobTitle, department, yearsOfService)
    {
        Mentor = mentor;
        StartDate = startDate;
        EndDate = endDate;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine( $"Наставник: {Mentor}, Начало стажировки: {StartDate.ToShortDateString()}, Конец стажировки: {EndDate.ToShortDateString()}");
    }

     public TimeSpan RemainingInternshipDuration()
    {
        return new DateTime(EndDate.Year, EndDate.Month, EndDate.Day) - DateTime.Now;
    }

}

// Интерфейс для управленческих обязанностей
public interface IManager
{
    void ManageTeam();
}

// Интерфейс для педагогических обязанностей
public interface IEducator
{
    void TeachClass();
}

public class Manager : Person, IManager, IEducator
{
    public string Title { get; set; }                 
    public int CountOfTeamMembers { get; set; } 

    public Manager(string name, int age, char gender, string email, long id, string citizenship, string title, int countOfTeamMembers) : base(name, age, gender, email, id, citizenship)
    {
        Title = title;
        CountOfTeamMembers = countOfTeamMembers;
    }

    public void ManageTeam()
    {
        Console.WriteLine( $"{Name} управляет командой из {CountOfTeamMembers} сотрудников.");
    }

    public void TeachClass()
    {
        Console.WriteLine( $"{Name} проводит занятие по управлению проектами.");
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine( $"Должность: {Title}");
    }
}


Person person = new Person("Иван Иванов", 30, 'М', "ivan@example.com", 1, "РФ");
Student student = new Student("Михаил Соколов", 20, 'М', "mikhail@example.com", 2, "РФ", "МФТИ", 3, "Компьютерные науки", 10000);
Employee employee = new Employee("Анна Петрова", 35, 'Ж', "anna@example.com", 3, "РФ", "Яндекс", 120000, "Менеджер проектов", "IT-отдел", 10);
Teacher teacher = new Teacher("Владимир Кузнецов", 45, 'М', "vladimir@example.com", 4, "РФ", "Математика", 20, "Высшая категория", true, 18);
Intern intern = new Intern("Кирилл Иванов", 25, 'М', "kirill@example.com", 5, "РФ", "Google", 50000, "Стажёр-программист", "Разработка ПО", 0, "Антон Федоров", new DateTime(2025, 9, 1),new DateTime(2025, 10, 25));
Manager manager = new Manager("Павел Сидоров", 40, 'М', "pavel@example.com", 6, "РФ", "Директор департамента", 3);


person.GetInfo();
person.IsAdult();
Console.WriteLine("");

student.GetInfo();
Console.WriteLine("");

employee.GetInfo();
Console.WriteLine( $"Размер премии Анны: {employee.CalculateBonus()} руб.");
Console.WriteLine("");

teacher.GetInfo();
Console.WriteLine( $"Годовая зарплата Владимира: {teacher.CalculateAnnualIncome(1000)} руб.");
Console.WriteLine("");

intern.GetInfo();
Console.WriteLine( $"Оставшийся срок стажировки сотрудника: {intern.RemainingInternshipDuration().Days} дней");
Console.WriteLine("");

manager.GetInfo();
manager.ManageTeam();
manager.TeachClass();
```
